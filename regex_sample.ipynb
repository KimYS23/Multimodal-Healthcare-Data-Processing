```python
import re

def normalize_loc(s: str) -> str:
    s = re.sub(r'\s+', ' ', s.strip())
    s = re.sub(r'\s*-\s*margin\b', '', s, flags=re.I)  # "- margin" 제거
    s = s.strip(' ,;.')
    return s

def parse_path(text: str) -> dict:
    out = {"Site": None, "수술위치": None, "수술방법": None}

    # 1) DIAGNOSIS 섹션 같은 한 줄 패턴: "Lung, ( right upper lobe, posterior segment ), segmentectomy:"
    m = re.search(
        r'(?im)^\s*(?:[A-Z]{1,3}(?:-[A-Z])?\)\s*)?(?P<site>Lung)\s*,\s*\(\s*(?P<loc>[^)]+?)\s*\)\s*,\s*(?P<proc>segmentectomy)\s*:',
        text,
        flags=re.I | re.M,
    )
    if m:
        out["Site"] = m.group("site").strip().title()
        out["수술위치"] = normalize_loc(m.group("loc"))
        out["수술방법"] = m.group("proc").strip().lower()
        return out

    # 2) Procedure 라인 보조
    m_proc = re.search(r'(?im)^\s*Procedure\s*:\s*(?P<proc>[^\n]+)', text)
    if m_proc:
        out["수술방법"] = m_proc.group("proc").strip().lower()

    # 3) Specimen 라인 보조: "Specimen: Lung (upper lobe) (posterior segment- margin)"
    m_spec = re.search(
        r'(?im)^\s*Specimen\s*:\s*(?P<site>Lung)\s*\(\s*(?P<lobe>[^)]+)\)\s*\(\s*(?P<seg>[^)]+)\)',
        text,
    )
    if m_spec:
        out["Site"] = m_spec.group("site").strip().title()
        lobe = normalize_loc(m_spec.group("lobe"))
        seg = normalize_loc(m_spec.group("seg"))
        out["수술위치"] = normalize_loc(f"{lobe}, {seg}")

    # 4) DIAGNOSIS 라인에서 Site/위치만 별도로 보조 추출 (콜론 없이 끝나는 경우 고려)
    if not out["Site"] or not out["수술위치"]:
        m_diag = re.search(
            r'(?im)\bLung\s*,\s*\(\s*(?P<loc>[^)]+?)\s*\)',
            text,
        )
        if m_diag:
            out["Site"] = out["Site"] or "Lung"
            out["수술위치"] = out["수술위치"] or normalize_loc(m_diag.group("loc"))

    return out
```

이 샘플에서 수술방법(surgical method) 은 "lobectomy"로 명확하게 명시되어 있습니다.
따라서 parse_path() 함수에서 "lobectomy"를 정규식으로 찾아내도록 수정하면 None이 아닌 "lobectomy"로 잘 인식됩니다. -> 여러 수술 키워드 패턴(PROC_PATTERN)

```python
import re

def normalize_loc(s: str) -> str:
    s = re.sub(r'\s+', ' ', s.strip())
    s = re.sub(r'\s*-\s*margin\b', '', s, flags=re.I)  # "- margin" 제거
    s = s.strip(' ,;.')
    return s

# 표준화 맵 + 키워드 패턴
PROC_ALIASES = {
    "lobectomy": "lobectomy",
    "sleeve lobectomy": "sleeve lobectomy",
    "bilobectomy": "bilobectomy",
    "segmentectomy": "segmentectomy",
    "wedge resection": "wedge resection",
    "sublobar resection": "sublobar resection",
    "pneumonectomy": "pneumonectomy",
    "resection": "resection",
    "biopsy": "biopsy",
}
PROC_PATTERN = r"(sleeve\s+lobectomy|bilobectomy|wedge\s+resection|sublobar\s+resection|pneumonectomy|segmentectomy|lobectomy|resection|biopsy)"

def standardize_proc(raw: str) -> str:
    if not raw:
        return ""
    key = raw.lower().strip()
    # 긴 키워드 우선
    for k in sorted(PROC_ALIASES, key=len, reverse=True):
        if re.search(rf"\b{re.escape(k)}\b", key):
            return PROC_ALIASES[k]
    return key

def parse_path(text: str) -> dict:
    out = {"Site": "", "수술위치": "", "수술방법": ""}

    # 1) DIAGNOSIS 한 줄 패턴: "Lung, ( left lower lobe ), lobectomy:"
    m = re.search(
        rf'(?im)^\s*(?:[A-Z]{{1,3}}(?:-[A-Z])?\)\s*)?'
        rf'(?P<site>Lung)\s*,\s*\(\s*(?P<loc>[^)]+?)\s*\)\s*,\s*(?P<proc>{PROC_PATTERN})\s*:',
        text,
        flags=re.I | re.M,
    )
    if m:
        out["Site"] = m.group("site").strip().title()
        out["수술위치"] = normalize_loc(m.group("loc"))
        out["수술방법"] = standardize_proc(m.group("proc"))
        return out

    # 2) Procedure: 라인 보조
    m_proc_line = re.search(r'(?im)^\s*Procedure\s*:\s*(?P<line>[^\n]+)', text)
    if m_proc_line:
        m_kw = re.search(rf'\b{PROC_PATTERN}\b', m_proc_line.group("line"), flags=re.I)
        out["수술방법"] = standardize_proc(m_kw.group(0) if m_kw else m_proc_line.group("line"))

    # 3) Specimen 라인 보조
    m_spec = re.search(
        r'(?im)^\s*Specimen\s*:\s*(?P<site>Lung)\s*\(\s*(?P<lobe>[^)]+)\)\s*\(\s*(?P<seg>[^)]+)\)',
        text,
    )
    if m_spec:
        out["Site"] = "Lung"
        lobe = normalize_loc(m_spec.group("lobe"))
        seg = normalize_loc(m_spec.group("seg"))
        out["수술위치"] = normalize_loc(f"{lobe}, {seg}")

    # 4) DIAGNOSIS 라인에서 위치만 보조
    if not out["Site"] or not out["수술위치"]:
        m_diag = re.search(r'(?im)\bLung\s*,\s*\(\s*(?P<loc>[^)]+?)\s*\)', text)
        if m_diag:
            out["Site"] = out["Site"] or "Lung"
            out["수술위치"] = out["수술위치"] or normalize_loc(m_diag.group("loc"))

    # 5) 전역 키워드 보조 (어디서든 수술명 발견되면 채움)
    if not out["수술방법"]:
        m_any = re.search(rf'\b{PROC_PATTERN}\b', text, flags=re.I)
        if m_any:
            out["수술방법"] = standardize_proc(m_any.group(0))

    # None/공백 방지
    for k in out:
        if out[k] is None:
            out[k] = ""
        out[k] = out[k].strip()

    return out
```
